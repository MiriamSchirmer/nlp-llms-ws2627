### 4.3 Subword tokens: how an LLM sees text

Language models process tokens, not words. Most of them use **subword tokenization** such as byte-pair encoding (BPE, see the slides): common words end up as single tokens, and rare words are split into smaller known pieces. Below we use `tiktoken`, the tokenizer library of OpenAI, with the vocabulary used by GPT-4 (`cl100k_base`). It needs an internet connection the first time, so we wrap it in `try` / `except`: if something fails, Python prints a message instead of stopping with an error.

```python
try:
    import tiktoken
    encoder = tiktoken.get_encoding("cl100k_base")    # the vocabulary used by GPT-4
    ids = encoder.encode("Unhappiness about the Bundestagswahl")
    pieces = []
    for i in ids:
        pieces.append(encoder.decode([i]))           # turn each number back into text
    print(ids)
    print(pieces)
except Exception as error:
    print("The subword tokenizer could not be loaded:", error)
```

The model only sees the numbers (token IDs). The space belongs to the token, so `' about'` and `'about'` are different tokens. Now the same for your text:

```python
try:
    ids = encoder.encode(my_text)
    pieces = []
    for i in ids:
        pieces.append(encoder.decode([i]))
    print(len(ids), "subword tokens for", len(tokens_split), "space-separated words")
    print(pieces[:40])
except Exception as error:
    print("The subword tokenizer is not available:", error)
```

```python
long_words = ["democracy", "Demokratie", "Bundesverfassungsgericht"]

try:
    for word in long_words:
        pieces = []
        for i in encoder.encode(word):
            pieces.append(encoder.decode([i]))
        print(word, "->", pieces)
except Exception as error:
    print("The subword tokenizer is not available:", error)
```

## 5. The pre-processing pipeline from the slides

We take the example sentence from the slides and apply one step after another. Most steps use the string methods from part 1. Step 3 uses `re.sub(pattern, replacement, text)`, which replaces everything that matches a **pattern**: `\d` means "a digit".

```python
text = "<div id=«text»> I’ve been in New York in 2011, but didn’t like it. I preferred LOS ANGELES.</div>"

# 1. Formatting removal: delete the HTML tags
step1 = text.replace("<div id=«text»>", "").replace("</div>", "").strip()
print("1. formatting removed:", step1)

# 2. Lower case
step2 = step1.lower()
print("2. lower case:        ", step2)
```

```python
# 3. Numbers normalization: every digit becomes 0
step3 = re.sub(r"\d", "0", step2)
print("3. numbers:           ", step3)

# 4. Join collocations: multi-word expressions become one token
step4 = step3.replace("new york", "new_york").replace("los angeles", "los_angeles")
print("4. collocations:      ", step4)
```

### 5.7 Stemming and lemmatization

Both reduce different forms of a word to one form:

- **Stemming** cuts off word endings with simple rules. It is fast but crude, and the result is not always a real word.
- **Lemmatization** finds the dictionary form (the *lemma*): *was* → *be*, *studies* → *study*. It needs a dictionary or a language model.

First a stemmer from NLTK (no download needed):

```python
from nltk.stem.snowball import SnowballStemmer

english_stemmer = SnowballStemmer("english")

for word in ["election", "elections", "elected", "taxes", "taxation", "studies"]:
    print(word, "->", english_stemmer.stem(word))
```

### 5.8 Putting it all together

Instead of running each step by hand, we put the steps into a **function**. `def` creates a function with a name; the indented lines are the steps; `return` gives back the result. We can then use the whole pipeline in one line, for any text.

```python
def preprocess(text, stopword_list):
    text = text.lower()                        # 1. lower case
    text = re.sub(r"[^\w\s]", "", text)        # 2. remove punctuation
    text = re.sub(r"\d+", "", text)            # 3. remove numbers (or use "0")
    result = []
    for token in text.split():                 # 4. tokenize
        if token not in stopword_list:         # 5. remove stopwords
            result.append(token)
    return result
```

## 6. Regular expressions: finding patterns

We have already used three patterns: `\w+`, `[^\w\s]` and `\d+`. A **regular expression** (regex) is a pattern that describes text. `re.findall(pattern, text)` returns every piece of the text that matches. Here are the examples from the slides:

```python
print(re.findall(r"Buttercup", "I’m called little Buttercup"))     # exact characters
print(re.findall(r"[mM]ary", "Mary and mary"))                     # [ ]: one of these characters
print(re.findall(r"[A-Z]", "Drenched Blossoms"))                   # a range: one capital letter
print(re.findall(r"[0-9]", "Chapter 1: Down the Rabbit Hole"))     # one digit
print(re.findall(r"cats?", "one cat, two cats"))                   # ?: the s is optional
print(re.findall(r"o+", "so sooooo"))                              # +: one or more
```

Two more that are useful for political texts:

```python
print(re.findall(r"[0-9]+", "In 2025, 3 parties won 120 seats."))   # one or more digits
print(re.findall(r"vot\w*", "vote, voters, voting, voted"))         # vot + any letters
```

**Word boundaries** (`\b`) make sure we match whole words only:

```python
s = "My cat is a catastrophe. He acts like a wildcat."

print(re.findall(r"cat", s))          # cat in cat, catastrophe and wildcat
print(re.findall(r"cat\b", s))        # cat and wildcat: the word must END after cat
print(re.findall(r"\bcat\b", s))      # only the full word cat
```

#### 📋 Regex cheat sheet

| Pattern | Matches | Example |
|---|---|---|
| `abc` | exactly these characters | `cat` finds *cat* in *concatenate* |
| `[abc]` | one of the characters in brackets | `[tT]he` → *the*, *The* |
| `[a-z]`, `[A-Z]`, `[0-9]` | one lowercase letter, capital letter, digit | |
| `\d` | a digit | `\d+` → *2025* |
| `\w` | a letter, digit or `_` | `\w+` → a whole word |
| `\s` | a space (or tab, line break) | |
| `[^...]` | any character **except** these | `[^\w\s]` → punctuation |
| `+` | the previous thing **one or more** times | `so+` → *sooooo* |
| `*` | the previous thing **zero or more** times | `vot\w*` → *vote*, *voters* |
| `?` | the previous thing is **optional** | `cats?` → *cat*, *cats* |
| `\b` | word boundary (start or end of a word) | `\bcat\b` → *cat* but not *wildcat* |

For POS tags, spaCy uses a statistical **model** that was trained on texts annotated by people. The results are the model's guesses, and they can be wrong.

First we choose the model for our language (English or German):

## 8. Dictionaries (lexicons)

A **lexicon** (or dictionary) is a list of words that stand for a concept: positive or negative sentiment, emotions, populism ... We count how many words of the lexicon appear in a text.

Here is a tiny sentiment lexicon in English and German. `+` joins two lists into one, so the lexicon works for texts in both languages.

```python
positive_en = ["good", "great", "helpful", "fair", "happy", "reliable", "clear", "like", "love", "kind"]
negative_en = ["bad", "terrible", "unfair", "unhappy", "expensive", "slow", "confusing", "problem", "angry"]
positive_de = ["gut", "toll", "hilfreich", "fair", "zufrieden", "wichtig", "klar"]
negative_de = ["schlecht", "schlimm", "unfair", "teuer", "langsam", "kritik", "problem"]

positive_words = positive_en + positive_de
negative_words = negative_en + negative_de
```

Now a function that counts: +1 for every positive word, -1 for every negative word.

```python
def sentiment_score(text):
    score = 0
    for token in re.findall(r"\w+", text.lower()):
        if token in positive_words:
            score = score + 1
        if token in negative_words:
            score = score - 1
    return score
```

```python
test_sentences = [
    "The debate was great and very helpful.",
    "The new law is bad for students.",
    "The plan is good, but the costs are terrible.",
]

for s in test_sentences:
    print(sentiment_score(s), "|", s)
```

Now your text, sentence by sentence. `re.split(r"[.!?]", my_text)` cuts the text at every `.`, `!` or `?`. We skip empty pieces.

```python
for s in re.split(r"[.!?]", my_text):
    if s.strip() != "":                       # .strip() removes spaces at the start and end
        print(sentiment_score(s), "|", s.replace("\n", " ").strip())     # line breaks become spaces
```

And the whole text. Longer texts contain more lexicon words, so we also compute the score **per 100 words** to compare texts of different length.

```python
n_words = len(re.findall(r"\w+", my_text))
score = sentiment_score(my_text)

print("Score:", score)
if n_words > 0:
    print("Score per 100 words:", round(100 * score / n_words, 1))
```

## 9. Sentiment with VADER: posts about the 2026 US midterms

Our lexicon from part 8 has fewer than 40 words. **VADER** (Hutto & Gilbert 2014) is a ready-made sentiment lexicon for social media text: a list of several thousand words whose sentiment was rated by people. On top of the word list it uses a few rules, for example for capital letters, exclamation marks, negation (*not good*) and words such as *very*. VADER only works for **English**.

VADER is part of NLTK. The lexicon has to be downloaded once.

```python
try:
    nltk.download("vader_lexicon", quiet=True)
    from nltk.sentiment.vader import SentimentIntensityAnalyzer
    vader = SentimentIntensityAnalyzer()
except Exception:
    print("The VADER lexicon could not be downloaded. Check the internet connection and run the cell again.")
```

`vader.polarity_scores(text)` returns four numbers:

- `neg`, `neu`, `pos`: the shares of the text that are negative, neutral and positive. Together they add up to 1.
- `compound`: one summary score from **-1** (very negative) to **+1** (very positive).

A common rule: a `compound` score of 0.05 or more counts as positive, -0.05 or less as negative, and everything in between as neutral.

```python
print(vader.polarity_scores("The debate was good."))
```

### VADER vs. our own lexicon

Our `sentiment_score()` from part 8 counts *good* the same way in all of these sentences. Compare it with the `compound` score from VADER:

```python
examples = [
    "The debate was good.",
    "The debate was GOOD.",
    "The debate was good!!!",
    "The debate was not good.",
    "Der Plan ist gut.",
]

for e in examples:
    print("our lexicon:", sentiment_score(e), "| VADER:", vader.polarity_scores(e)["compound"], "|", e)
```

Capital letters and exclamation marks make the VADER score stronger, and *not* turns it negative. The German sentence gets a VADER score of 0, because VADER does not know German words (our own lexicon does, because we added them).

### VADER scores for every post

We compute the `compound` score for every post and turn it into a label with the rule from above. The new lists become new columns of the table.

```python
def vader_label(score):
    if score >= 0.05:
        return "positive"
    elif score <= -0.05:
        return "negative"
    else:
        return "neutral"
```

```python
scores = []
labels = []
for text in posts["text"]:
    compound = vader.polarity_scores(text)["compound"]
    scores.append(compound)
    labels.append(vader_label(compound))

posts["compound"] = scores
posts["vader_label"] = labels
posts[["name", "party", "text", "compound", "vader_label"]]
```

### A chart of the scores

One bar per post, sorted by score: blue for Democrats, red for Republicans. Posts with a score of 0 have no visible bar. First we prepare the labels and colours:

```python
sorted_posts = posts.sort_values("compound")     # from the lowest to the highest score

bar_labels = []
bar_colors = []
for i in sorted_posts.index:
    bar_labels.append(sorted_posts.loc[i, "name"] + ": " + sorted_posts.loc[i, "text"][:35])
    if sorted_posts.loc[i, "party"] == "Democrat":
        bar_colors.append("#2E5EAA")     # blue
    else:
        bar_colors.append("#C0392B")     # red
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 8))
plt.barh(bar_labels, sorted_posts["compound"], color=bar_colors)
plt.axvline(0, color="grey", linewidth=0.8)     # a line at 0
plt.xlim(-1, 1)
plt.xlabel("VADER compound score (blue = Democrat, red = Republican)")
plt.title("VADER sentiment of posts after the 2026 midterms")
plt.tight_layout()
plt.show()
```

And the mean score per party. `groupby("party")` splits the table by party, and `.mean()` computes the average of each group.

```python
means = posts.groupby("party")["compound"].mean()
print(means.round(3))

plt.figure(figsize=(4, 3))
plt.bar(means.index, means.values, color=["#2E5EAA", "#C0392B"])
plt.axhline(0, color="grey", linewidth=0.8)
plt.ylabel("Mean compound score")
plt.title("Mean VADER score by party")
plt.tight_layout()
plt.show()
```
